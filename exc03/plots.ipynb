```python
import pandas as pd
import matplotlib.pyplot as plt
import io
```

```python

# --- falls du die Daten direkt im Code einfügen willst ---
data = """2 83279058.939391
8 329921818.758464
64 2221709946.437047
512 9971243818.942564
4096 171224036577.400391
32768 845625806451.612915
262144 4369066666666.666992
2097152 34952533333333.335938
16777216 279620266666666.687500
134217728 1491308088888889.000000
"""
# Wenn du eine Datei hast: df = pd.read_csv("data.csv", sep="\s+", header=None)
df = pd.read_csv(io.StringIO(data), sep=r"\s+", header=None, names=["size", "throughput"])

# Alle negativen oder Nullwerte entfernen (Sicherheitscheck)
df = df[(df["size"] > 0) & (df["throughput"] > 0)]

plt.figure(figsize=(8,5))
plt.plot(df["size"], df["throughput"], linestyle="-", color="b")

plt.xscale("log")
plt.yscale("log")

plt.xlabel("Data size")
plt.ylabel("Throughput (Bytes/s)")
plt.title("GPU Memory Copy Throughput")
plt.grid(True, which="both", ls="--", lw=0.5)
plt.tight_layout()
plt.show()
```
