```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
d2d = pd.read_csv("D2D-data-transfer.csv", 
                delim_whitespace=True,
                comment='M',
                skiprows=1, 
                names=["data", "time", "transfer"])
```

```python
plt.plot(d2d['data'][:10], d2d['transfer'][:10])

plt.title('Data transfer in device')
plt.xlabel('data size [bytes]')
plt.ylabel('transfer rate [bytes/s]')
plt.savefig('d2d_plot.png')
plt.show()
```
