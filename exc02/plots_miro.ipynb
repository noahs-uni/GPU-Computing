```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
breakeven = pd.read_csv("Break-even_startup_miro.txt", 
                 delim_whitespace=True, 
                 skiprows=1, 
                 names=["cycles", "time"])
```

```python
plt.plot(breakeven['cycles'][:80], breakeven['time'][:80])
plt.title('Break-even time for kernel startup')
plt.xlabel('clock cycles')
plt.ylabel(r'kernel startup time [$\mu$s]')
plt.savefig('break-even.png')
plt.show()
```

```python
d2h = pd.read_csv("Data_transfer_D2H_miro.txt", 
                 delim_whitespace=True,
                    comment='M',
                 skiprows=1, 
                 names=["data", "time", "transfer"])

h2d = pd.read_csv("Data_transfer_H2D_miro.txt", 
                 delim_whitespace=True,
                  comment='M',
                 skiprows=1, 
                 names=["data", "time", "transfer"])
```

```python
plt.plot(d2h['data'][:10], d2h['transfer'][:10], label='pageable mem.')
plt.plot(d2h['data'][10:], d2h['transfer'][10:], label='pinned mem.')
plt.xscale('log')
plt.yscale('log')
plt.title('Data transfer device to host')
plt.xlabel('data size [bytes]')
plt.ylabel('transfer rate [bytes/s]')
plt.legend()
plt.savefig('data_d2h.png')
plt.show()
```

```python
plt.plot(h2d['data'][:10], h2d['transfer'][:10], label='pageable mem.')
plt.plot(h2d['data'][10:], h2d['transfer'][10:], label='pinned mem.')
plt.xscale('log')
plt.yscale('log')
plt.title('Data transfer host to device')
plt.xlabel('data size [bytes]')
plt.ylabel('transfer rate [bytes/s]')
plt.legend()
plt.savefig('data_h2d.png')
plt.show()
```
